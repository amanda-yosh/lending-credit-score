```python
import io
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.patches import Patch

# 1. Dados da Query Q1.4
csv_q1_4 = """project,blockchain,n_borrowers,n_borrows,borrows_por_wallet
aave,polygon,185292,4393998,23.71
aave,ethereum,122393,1175253,9.60
aave,arbitrum,196077,1542016,7.86
aave,base,196496,1751518,8.91
compound,ethereum,22736,185548,8.16
compound,base,13996,95572,6.83
compound,arbitrum,6456,47285,7.32
compound,polygon,1500,13164,8.78
"""

df = pd.read_csv(io.StringIO(csv_q1_4))

# 2. Preparação do DataFrame
df_sorted = df.sort_values(by='borrows_por_wallet', ascending=True).reset_index(drop=True)
df_sorted['label'] = df_sorted['project'].str.capitalize() + ' (' + df_sorted['blockchain'].str.capitalize() + ')'

# 3. Estilo e Figura
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 8))

color_map = {'aave': '#377eb8', 'compound': '#ff7f00'}
colors = [color_map[p] for p in df_sorted['project']]

# 4. Desenho das Barras
bars = ax.barh(df_sorted['label'], df_sorted['borrows_por_wallet'], color=colors, alpha=0.85, height=0.65)

# Rótulos de valor ao lado de cada barra
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.3, bar.get_y() + bar.get_height()/2, f'{width:.2f}',
            va='center', ha='left', fontsize=8.5, fontweight='bold', color='#333333')

# 5. Formatação dos Eixos e Legenda
ax.set_xlim(0, 27) # Espaço extra para o texto das últimas barras
ax.set_title('Q1.4 — Sequence Richness: Average Borrows per Wallet by Chain & Protocol',
             fontsize=12, fontweight='bold', pad=15)
ax.set_xlabel('Average Borrow Events per Wallet (Borrows / Wallet)', fontsize=10, fontweight='bold')
ax.set_ylabel('')

legend_elements = [
    Patch(facecolor='#377eb8', alpha=0.85, label='Aave'),
    Patch(facecolor='#ff7f00', alpha=0.85, label='Compound')
]
ax.legend(handles=legend_elements, loc='lower right', frameon=True, facecolor='white', framealpha=0.9, edgecolor='lightgray')

plt.tight_layout()

# Salvar para o Overleaf
plt.savefig('Q1_4_borrows_por_wallet.pdf', dpi=300, bbox_inches='tight')
plt.savefig('Q1_4_borrows_por_wallet.png', dpi=300, bbox_inches='tight')
plt.show()
```
