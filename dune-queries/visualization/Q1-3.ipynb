```python
import matplotlib.pyplot as plt
import seaborn as sns

# Dados da Query Q1.3
total_par = 20585
simultaneo = 9428
migracao = total_par - simultaneo

pct_simultaneo = (simultaneo / total_par) * 100
pct_migracao = (migracao / total_par) * 100

# Estilo acadêmico
sns.set_theme(style="white")
fig, ax = plt.subplots(figsize=(6, 5))

# Cores: Verde para Multi-homing ativo, Azul neutro para Migração
colors = ['#2ca02c', '#1f77b4']
labels = [
    f'Active Multi-homing\n(Simultaneous Active Windows)\n{simultaneo:,} ({pct_simultaneo:.1f}%)',
    f'Sequential Migration\n(No Time Overlap)\n{migracao:,} ({pct_migracao:.1f}%)'
]

# Gráfico de Donut
wedges, texts, autotexts = ax.pie(
    [simultaneo, migracao],
    labels=labels,
    autopct='',
    startangle=140,
    colors=colors,
    wedgeprops=dict(width=0.4, edgecolor='w', linewidth=2)
)

# Estilização das fontes
for t in texts:
    t.set_fontsize(9.5)
    t.set_fontweight('bold')

plt.title('Q1.3 — Temporal Overlap of Multi-Protocol Borrowers\n(Aave vs. Compound)',
          fontsize=11, fontweight='bold', pad=20)

plt.tight_layout()
plt.savefig('Q1_3_temporal_overlap.pdf', dpi=300, bbox_inches='tight')
plt.savefig('Q1_3_temporal_overlap.png', dpi=300, bbox_inches='tight')
plt.show()
```
