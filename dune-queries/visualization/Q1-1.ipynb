```python
import io
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.dates as mdates
from matplotlib.patches import Patch

# 1. Dados da Query Q1.1
csv_q1_1 = """project,version,blockchain,n_events,n_borrowers,first,last
aave,3,base,1752735,196565,2023-08-22 15:12:27,2026-07-26 23:24:53
aave,3,arbitrum,1543168,196134,2022-03-16 16:11:55,2026-07-26 23:31:43
aave,3,polygon,1420878,110210,2022-03-14 12:40:09,2026-07-26 23:19:27
aave,3,optimism,862500,98764,2022-03-15 21:49:16,2026-07-26 23:19:41
aave,2,polygon,2973336,94907,2021-03-31 11:16:05,2025-03-06 18:49:13
aave,3,ethereum,820732,90510,2023-01-27 08:24:47,2026-07-26 23:29:11
aave,3,scroll,266364,70237,2024-02-09 18:50:42,2026-04-10 01:06:51
aave,3,avalanche_c,529227,61194,2022-03-16 20:42:15,2026-07-26 23:26:27
aave,2,ethereum,271578,51288,2020-12-01 16:46:58,2025-03-10 08:46:23
aave,3,bnb,412789,19730,2024-01-23 15:07:23,2026-07-26 23:21:14
aave,2,avalanche_c,325905,18235,2021-10-04 13:25:54,2025-06-26 21:28:18
aave,3,linea,82352,11461,2025-02-11 12:15:29,2026-07-26 18:40:15
aave,1,ethereum,83653,10787,2020-01-07 08:56:10,2022-12-21 18:54:23
aave,3,sonic,93058,7973,2025-03-03 19:00:23,2026-06-30 14:09:07
aave,3,zksync,19402,4966,2024-09-20 14:11:15,2026-02-11 14:05:39
aave,3,plasma,36024,4386,2025-09-24 14:14:58,2026-07-26 22:04:52
aave,3,gnosis,73084,4165,2023-11-07 09:17:50,2026-07-26 20:32:05
aave,3,celo,59305,3013,2025-03-17 13:46:15,2026-07-26 22:00:05
aave,3,fantom,11591,1434,2022-03-16 17:15:10,2022-09-16 08:16:27
compound,2,ethereum,118968,19494,2019-05-07 02:48:28,2025-12-08 10:45:47
compound,3,base,96192,14011,2023-08-12 17:44:15,2026-07-26 23:35:29
compound,3,arbitrum,47305,6459,2023-05-16 01:21:39,2026-07-26 19:59:02
compound,3,ethereum,63034,5949,2022-08-26 01:11:09,2026-07-26 22:33:47
compound,3,unichain,6855,3674,2025-03-10 01:29:36,2026-07-06 15:33:57
compound,3,polygon,13164,1500,2023-03-07 18:57:02,2026-07-22 06:54:00
compound,1,ethereum,3578,632,2018-09-19 02:48:02,2026-04-25 10:47:23"""

df = pd.read_csv(io.StringIO(csv_q1_1))

# 2. Tratamento de datas e rótulos
df['first'] = pd.to_datetime(df['first'])
df['last'] = pd.to_datetime(df['last'])
df['label'] = df['project'].str.capitalize() + ' v' + df['version'].astype(str) + ' (' + df['blockchain'].str.capitalize() + ')'

# Ordenar por número de mutuários
df = df.sort_values(by='n_borrowers', ascending=True).reset_index(drop=True)

# 3. Configuração do gráfico
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(11, 8))

color_map = {'aave': '#377eb8', 'compound': '#ff7f00'}

# 4. Construção das Barras Horizontais
for idx, row in df.iterrows():
    start_date = row['first']
    end_date = row['last']
    duration_days = (end_date - start_date).days

    ax.barh(
        y=row['label'],
        width=duration_days,
        left=start_date,
        color=color_map[row['project']],
        alpha=0.75,
        height=0.55
    )

    # Texto ao lado da barra
    n_borrowers_str = f" {row['n_borrowers']:,} wallets"
    ax.text(
        end_date,
        idx,
        n_borrowers_str,
        va='center',
        ha='left',
        fontsize=8,
        fontweight='bold',
        color='#333333'
    )

# 5. Formatação do Eixo X e Posição da Legenda
ax.xaxis.set_major_locator(mdates.YearLocator(1))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

# Expandido até 2028.5 para dar folga aos textos
ax.set_xlim(pd.to_datetime('2018-01-01'), pd.to_datetime('2028-06-01'))

ax.set_title('Q1.1 — Temporal Coverage and Active Borrower Volume by Protocol, Version & Chain',
             fontsize=12, fontweight='bold', pad=15)
ax.set_xlabel('Historical Window (2018 – 2026)', fontsize=10, fontweight='bold')
ax.set_ylabel('')

# Legenda movida para o topo esquerdo (espaço vazio e limpo do gráfico)
legend_elements = [
    Patch(facecolor='#377eb8', alpha=0.75, label='Aave Deployment'),
    Patch(facecolor='#ff7f00', alpha=0.75, label='Compound Deployment')
]
ax.legend(handles=legend_elements, loc='upper left', frameon=True, facecolor='white', framealpha=0.9, edgecolor='lightgray')

plt.tight_layout()

# Salvar para o Overleaf
plt.savefig('Q1_1_cobertura_temporal.pdf', dpi=300, bbox_inches='tight')
plt.savefig('Q1_1_cobertura_temporal.png', dpi=300, bbox_inches='tight')
plt.show()
```
